```python
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# Create custom legend entries
black_dashed_line = mlines.Line2D([], [], color='black', linestyle='--', label='Ground Truth')
start_point = mlines.Line2D([], [], color='black', marker='o', linestyle='None', label='Start Point')
end_point = mlines.Line2D([], [], color='black', marker='x', linestyle='None', label='End Point')
color_map = cm.ScalarMappable(cmap='viridis', norm=mcolors.Normalize(vmin=0, vmax=1))
color_map._A = []  # Suppress the warning by setting the array

# Create a dummy line to represent the color map in the legend
color_map_entry = mlines.Line2D([0,1,2], [0,1,2], color=color_map.to_rgba(0.5), marker='None', linestyle='-', label='Trajectroy of Interest, Colormap: ||trans(APE)||')

# Create a figure just for the legend
fig, ax = plt.subplots(figsize=(4, 1))
ax.axis('off')  # Hide axes

# Add legend
legend = fig.legend(handles=[black_dashed_line, start_point, end_point, color_map_entry], loc='center')

# Show the legend-only plot
plt.show()

fig.savefig("legend.pdf")
```
