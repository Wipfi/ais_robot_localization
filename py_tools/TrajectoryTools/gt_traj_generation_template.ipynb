# Load Trajectory

```python
from evo.core.trajectory import PosePath3D, PoseTrajectory3D
from scipy.spatial.transform import Rotation as Rot
import numpy as np
import dill

dataset = "10"
folder = "Sets"

dumpfile = open(f"../notebooks/Dataset_Syntetic/TrajectoryGeneratorOutput/rand{dataset}.dill", "rb")
trajectory: PoseTrajectory3D = dill.load(dumpfile)
dumpfile.close()
```

# Set Origin To Zero

```python
trajectory.transform(np.linalg.inv(trajectory.poses_se3[0]))

del trajectory.meta["base_points"]
```

# Save GT

```python
trajectory.meta["Name"] = "gt"
trajectory.meta["topic"] = "/ground_truth"
trajectory.meta["frame_id"] = "map"
trajectory.meta["child_frame_id"] = "base_link"
trajectory.meta["publish_transform"] = False
trajectory.meta["publish_orientation"] = False
trajectory.meta["frequency"] = 10 

# Save the new trajectory
output_file = f"../notebooks/Dataset_Syntetic/{folder}/{dataset}/gt.dill"
with open(output_file, "wb") as f:
    dill.dump(trajectory, f)

print(f"Drifted trajectory with linear and angular drift (per-axis) saved to {output_file}")

#trajectory.meta["frequency"] = 5 

# Save the new trajectory
#output_file = "../notebooks/Dataset_Syntetic/Sets/1/gt_half_speed.dill"
#with open(output_file, "wb") as f:
    #dill.dump(trajectory, f)

#print(f"Drifted trajectory with linear and angular drift (per-axis) saved to {output_file}")
```

#Optinal: Assume perfect GNSS
